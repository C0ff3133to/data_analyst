# Nested For Loops

```python
flavors = ['Vanilla', 'Chocolate', 'Cookie Dough']
toppings = ['Hot fudge', 'Oreos', 'Marshmallows']
```

```python
for one in flavors:
    for two in toppings:
        print()
```
