# if - elif - else statements

```python
if 25 > 10:
    print('it worked!')
```

```python
if 25 < 10:
    print('it worked!')
else:
    print('It did not work...')
```

```python
# elif 3 worked! will also run but a way to test so multiple runs.
```

```python
if (25 < 10) or (1 < 3):
    print('it worked!')
elif 25 < 20:
    print('elif worked!')
elif 25 < 21:
    print('elif 2 worked!')
elif 25 < 40:
    print('elif 3 worked!')
elif 25 < 50:
    print('elif 4 worked!')
else:
    print('It did not work...')
```

```python
print('It worked!') if 10>30 else print ('It did not work...')
```

```python
# Nested

if (25 < 10) or (1 < 3):
    print('it worked!')
    if 10 > 5:
        print('This nested if statement worked!')
elif 25 < 20:
    print('elif worked!')
elif 25 < 21:
    print('elif 2 worked!')
elif 25 < 40:
    print('elif 3 worked!')
elif 25 < 50:
    print('elif 4 worked!')
else:
    print('It did not work...')
```
