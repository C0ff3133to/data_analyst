# Merge, Join & Concatenate

# inner being natural join
# outer being full outer join
# left being left outer join
# right being right outer join

```python
import pandas as pd
```

```python
df1 = pd.read_csv(r"C:\Users\wijks\Documents\Data Analyst Portfolio\Excel\LOTR.csv")
df1
```

```python
df2 = pd.read_csv(r"C:\Users\wijks\Documents\Data Analyst Portfolio\Excel\LOTR 2.csv")
df2
```

```python
# default an inner join
# gives where specific values or keys are the same
df1.merge(df2)
```

```python
# inner
df1.merge(df2, how = 'inner', on = ['FellowshipID', 'FirstName'])
```

```python
# outer
df1.merge(df2, how = 'outer')
```

```python
# left
df1.merge(df2, how = 'left')
```

```python
# right
df1.merge(df2, how = 'right')
```

```python
# cross
df1.merge(df2, how = 'cross')
```

```python
# joins
df1.join(df2, on = 'FellowshipID', how = 'outer', lsuffix = '_Left',rsuffix = '_Right')
```

```python
df4 = df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'), lsuffix = '_Left',rsuffix = '_Right', how = 'outer')
df4
```

```python
# concatenate = putting one data frame on top of the other instead of next to another df
```

```python
pd.concat([df1,df2], join = 'outer', axis = 1)
```
