# Pandas Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\wijks\Documents\Data Analyst Portfolio\Excel\Ice Cream Ratings.csv")
df = df.set_index('Date')
df
```

```python
df.plot(kind = 'line', subplots = True)
```

```python
# Styles available for styling
print(plt.style.available)
plt.style.use('seaborn-v0_8-bright')
```

```python
# set title
df.plot(kind = 'line', title = 'Ice Cream Ratings')
```

```python
# set axis names
df.plot(kind = 'line', title = 'Ice Cream Ratings', xlabel = 'Daily Ratings', ylabel = 'Scores')
```

```python
df.plot(kind = 'bar', stacked = True)
```

```python
df['Flavor Rating'].plot(kind = 'bar', stacked = True)
```

```python
# horizontal
df.plot.barh(stacked = True)
```

```python
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating')
```

```python
# options for scatter plots
# size
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating', s = 100)
```

```python
# change colors
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating', c = 'red')
```

```python
# histogram
df.plot.hist(bins = 10)
```

```python
# box plot
df.boxplot()
```

```python
# area plot
df.plot.area(figsize = (10,5))
```

```python
# pie chart
df.plot.pie(y = 'Flavor Rating', figsize = (26,11))
```
