# Automatic File Sorter in File Explorer

```python
import os, shutil
```

```python
path = r"C:/Users/wijks/Documents/wijk studio/"
```

```python
file_name = os.listdir(path)
```

```python
# Check if a format in the path do exist or not.
os.path.exists(path + 'csv files')
```

```python
# Creation of maps if not existing yet.
folder_names = ["csv files", "image files", "text files", "pdf files"]

for folder in folder_names:
    folder_path = os.path.join(path, folder)
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
        print(f"Created the map: {folder}")

# Sorting of files
for file in os.listdir(path):
    file_path = os.path.join(path, file)
    
    # Skip of files
    if os.path.isdir(file_path):
        continue
    
    # Moves files depending on format
    if file.lower().endswith('.csv'):
        shutil.move(file_path, os.path.join(path, "csv files", file))
        
    elif file.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')):
        shutil.move(file_path, os.path.join(path, "image files", file))

    elif file.lower().endswith(('.pdf')):
        shutil.move(file_path, os.path.join(path, "pdf files", file))
        
    elif file.lower().endswith('.txt'):
        shutil.move(file_path, os.path.join(path, "text files", file))
```
